```python
import random
import time
```

```python
class Personagem:
    def __init__(self, nome, hp, mp, speed, atk, matk, defense, mdefense):
        self.nome = nome
        self.hp = hp
        self.max_hp = hp
        self.mp = mp
        self.max_mp = mp
        self.speed = speed
        self.atk = atk
        self.matk = matk
        self.defense = defense
        self.mdefense = mdefense

    def esta_vivo(self):
        return self.hp > 0

    def receber_dano(self, dano):
        dano_real = max(1, dano)  # Garante pelo menos 1 de dano
        self.hp = max(0, self.hp - dano_real)
        return dano_real

    def atacar(self, alvo):
        """Método base a ser sobrescrito pelas classes filhas (Polimorfismo)."""
        dano = self.atk - alvo.defense
        dano_causado = alvo.receber_dano(dano)
        print(f"⚔️ {self.nome} ataca {alvo.nome} causando {dano_causado} de dano físico!")
```

```python
class Guerreiro(Personagem):
    def __init__(self, nome):
        # hp, mp, speed, atk, matk, defense, mdefense
        super().__init__(nome, hp=150, mp=30, speed=10, atk=25, matk=5, defense=15, mdefense=8)

    def atacar(self, alvo):
        # Habilidade especial: Golpe Poderoso se tiver Mana suficiente
        if self.mp >= 10:
            self.mp -= 10
            dano = (self.atk * 1.5) - alvo.defense
            dano_causado = alvo.receber_dano(dano)
            print(f"🪓 {self.nome} usa [Golpe Pesado] em {alvo.nome} causando {dano_causado} de dano! (MP restante: {self.mp})")
        else:
            dano = self.atk - alvo.defense
            dano_causado = alvo.receber_dano(dano)
            print(f"⚔️ {self.nome} realiza um ataque básico em {alvo.nome} causando {dano_causado} de dano!")

class Mago(Personagem):
    def __init__(self, nome):
        super().__init__(nome, hp=80, mp=100, speed=12, atk=8, matk=30, defense=5, mdefense=15)

    def atacar(self, alvo):
        # Utiliza ataque mágico (matk vs mdefense)
        if self.mp >= 15:
            self.mp -= 15
            dano = self.matk - alvo.mdefense
            dano_causado = alvo.receber_dano(dano)
            print(f"🔥 {self.nome} lança [Bola de Fogo] em {alvo.nome} causando {dano_causado} de dano mágico! (MP restante: {self.mp})")
        else:
            dano = self.atk - alvo.defense
            dano_causado = alvo.receber_dano(dano)
            print(f"🪄 {self.nome} acerta {alvo.nome} com o cajado causando {dano_causado} de dano físico!")

class Paladino(Personagem):
    """Classe inédita com mecânica de cura e ataque sagrado misto."""
    def __init__(self, nome):
        super().__init__(nome, hp=130, mp=60, speed=8, atk=18, matk=18, defense=12, mdefense=12)

    def atacar(self, alvo):
        # Polimorfismo: Ataque Sagrado que usa média de ATK + MATK e cura a si mesmo
        if self.mp >= 12:
            self.mp -= 12
            dano_base = ((self.atk + self.matk) / 2) - ((alvo.defense + alvo.mdefense) / 2)
            dano_causado = alvo.receber_dano(dano_base)

            # Cura 10% da vida máxima ao atacar
            cura = int(self.max_hp * 0.10)
            self.hp = min(self.max_hp, self.hp + cura)
            print(f"✨ {self.nome} usa [Julgamento Sagrado] em {alvo.nome} causando {dano_causado} de dano e recupera {cura} de HP!")
        else:
            dano = self.atk - alvo.defense
            dano_causado = alvo.receber_dano(dano)
            print(f"⚔️ {self.nome} ataca {alvo.nome} com sua espada causando {dano_causado} de dano!")
```

```python
class Monstro:
    def __init__(self, nome, hp, mp, speed, atk, matk, defense, mdefense):
        self.nome = nome
        self.hp = hp
        self.max_hp = hp
        self.mp = mp
        self.max_mp = mp
        self.speed = speed
        self.atk = atk
        self.matk = matk
        self.defense = defense
        self.mdefense = mdefense

    def esta_vivo(self):
        return self.hp > 0

    def receber_dano(self, dano):
        dano_real = max(1, dano)
        self.hp = max(0, self.hp - dano_real)
        return dano_real

    def atacar(self, alvo):
        # Ataque padrão escolhe aleatoriamente entre físico e mágico se tiver MP
        if self.matk > self.atk and self.mp >= 5:
            self.mp -= 5
            dano = self.matk - alvo.mdefense
            dano_causado = alvo.receber_dano(dano)
            print(f"👾 {self.nome} lança um ataque sombrio em {alvo.nome} causando {dano_causado} de dano mágico!")
        else:
            dano = self.atk - alvo.defense
            dano_causado = alvo.receber_dano(dano)
            print(f"👹 {self.nome} morde/golpea {alvo.nome} causando {dano_causado} de dano físico!")
```

```python
def batalha_1v1(heroi, monstro):
    print(f"\n==========================================")
    print(f"⚔️  INÍCIO DO COMBATE 1v1: {heroi.nome} vs {monstro.nome}  ⚔️")
    print(f"==========================================")

    turno = 1
    # Define a ordem de quem ataca primeiro pela SPEED
    if heroi.speed >= monstro.speed:
        primeiro, segundo = heroi, monstro
    else:
        primeiro, segundo = monstro, heroi

    print(f"⚡ {primeiro.nome} é mais rápido (Speed: {primeiro.speed}) e ataca primeiro!\n")

    while heroi.esta_vivo() and monstro.esta_vivo():
        print(f"--- TURNO {turno} ---")

        # Primeiro a agir
        primeiro.atacar(segundo)
        if not segundo.esta_vivo():
            print(f"\n☠️  {segundo.nome} foi derrotado!")
            break

        # Segundo a agir
        segundo.atacar(primeiro)
        if not primeiro.esta_vivo():
            print(f"\n☠️  {primeiro.nome} foi derrotado!")
            break

        print(f"Status: {heroi.nome} [HP: {heroi.hp}/{heroi.max_hp}] | {monstro.nome} [HP: {monstro.hp}/{monstro.max_hp}]\n")
        turno += 1

    vencedor = heroi if heroi.esta_vivo() else monstro
    print(f"\n🏆 VITORIA DE {vencedor.nome.upper()}!")

def batalha_3v3(time_herois, time_monstros):
    print(f"\n==========================================")
    print(f"🛡️  INÍCIO DO COMBATE EM EQUIPE 3v3  🐉")
    print(f"==========================================")

    turno = 1

    while any(h.esta_vivo() for h in time_herois) and any(m.esta_vivo() for m in time_monstros):
        print(f"\n--- TURNO {turno} ---")

        # Junta todos os combatentes vivos e ordena por SPEED (maior para menor)
        todos_combatentes = [c for c in time_herois + time_monstros if c.esta_vivo()]
        todos_combatentes.sort(key=lambda c: c.speed, reverse=True)

        for combatente in todos_combatentes:
            if not combatente.esta_vivo():
                continue

            # Determina se é Herói ou Monstro para escolher o time rival
            if combatente in time_herois:
                alvos_validos = [m for m in time_monstros if m.esta_vivo()]
                if not alvos_validos:
                    break
                alvo = random.choice(alvos_validos)
            else:
                alvos_validos = [h for h in time_herois if h.esta_vivo()]
                if not alvos_validos:
                    break
                alvo = random.choice(alvos_validos)

            combatente.atacar(alvo)

            if not alvo.esta_vivo():
                print(f"   ☠️ {alvo.nome} caiu em batalha!")

        # Exibe status da rodada
        print("\nStatus da Equipe de Heróis:")
        for h in time_herois:
            status = f"{h.hp}/{h.max_hp} HP" if h.esta_vivo() else "FORA DE COMBATE"
            print(f" - {h.nome}: {status}")

        print("Status da Equipe de Monstros:")
        for m in time_monstros:
            status = f"{m.hp}/{m.max_hp} HP" if m.esta_vivo() else "FORA DE COMBATE"
            print(f" - {m.nome}: {status}")

        turno += 1

    # Resultado do Combate 3v3
    if any(h.esta_vivo() for h in time_herois):
        print(f"\n🎉 A EQUIPE DOS HERÓIS VENCEU O COMBATE EM {turno - 1} TURNOS!")
    else:
        print(f"\n💀 A EQUIPE DOS MONSTROS VENCEU O COMBATE EM {turno - 1} TURNOS!")
```
